### pixels and image matrix representation

- it is smallest unit of image
- eg 4*4 images

100 | 121 | 99 | 61
-

-
109 | 124 | 86 | 63
-
110 | 123 | 66 | 64

- each number represents intensity of image

### resolution 

- resolution = width * height

### Color model 

RGB : Red Green Blue

| R | G | B | Color |
| -- | -- | -- | -- |
| 255 | 0 | 0 | Red |
| 0 | 255 | 0 | Green |
| 0 | 0 | 255 | Blue |
| 0 | 0 | 0 | Black |
| 255 | 255 | 255 | White |


### HSV Model : 
- Hue: Base color type (RGB)

```python
## resizing with keeping aspect ratio

height, width = img.shape[:2]

new_width = width//2
new_height = height//2

resized_new = cv2.resize(
    img,
    (new_width, new_height)
)

cv2.imshow("resized image", resized_new)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
## different interpolation methods

Resize = cv2.resize(
    img,
    (300,300), interpolation = cv2.INTER_AREA
)

cv2.imshow("Inter area",Resize)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
### rotate it as you like 

img1 = cv2.imread("ronaldo.jpg")
height, width = img.shape[:2]
center = (width//2, height//2)

matrix = cv2.getRotationMatrix2D(
    center,
    180,
    1.0
)

rotated = cv2.warpAffine(
    img1,
    matrix,
    (width,height)
)

cv2.imshow("rotate",rotated)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
## flip 

## horizontal flip

flipped = cv2.flip(
    img1,
    1
)

cv2.imshow("flip the image",flipped)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
## vertical flip

flipped = cv2.flip(
    img1,
    -1
)

cv2.imshow("flip the image",flipped)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
## translation

## moving image position

## translation matrix

import cv2
import numpy as np

img = cv2.imread("hitler.jpg")

rows, coln = img.shape[:2]

M = np.float32(
    [
        [1,0,100],
        [0,1,50]
    ]
)

translated = cv2.warpAffine(
    img,
    M,
    (coln, rows)
)

cv2.imshow("traslated image of hitler", translated)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
### Scaling

scaled = cv2.resize(
    img,
    None,
    fx = 0.5,
    fy = 0.5,
)

cv2.imshow("scaled image of hitler", scaled)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### histogram analysis

- A histogram shows the distribution of pixels intensity
- For gray-scale image:

X-axis = Intensity(0-255)
Y-axis = number of pixels

```python
import matplotlib.pyplot as plt

img_grey = cv2.imread("Balen.webp",0)

hist = cv2.calcHist([img_grey],[0], None, [256],[0,256])

plt.plot(hist)
plt.title("Histogram")
plt.xlabel("Intensity")
plt.ylabel("Number of pixel")

plt.show()
```

```python
## histogram qulaity enhancement

equalized = cv2.equalizeHist(img_grey)
```

```python
cv2.imshow("Messi", img_grey)
cv2.waitKey(0)
cv2.destroyAllWindows()

cv2.imshow("Messi", equalized)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
hist = cv2.calcHist([equalized],[0], None, [256],[0,256])

plt.plot(hist)
plt.title("Histogram")
plt.xlabel("Intensity")
plt.ylabel("Number of pixel")

plt.show()
```

```python
## image smoothing
img_blur = cv2.blur(img, (5,5))
cv2.imshow("Messi", img_blur)
cv2.waitKey(0)
cv2.destroyAllWindows()

## gaussian blur
gaussian_blur = cv2.GaussianBlur(img,(5,5),0)
cv2.imshow("Messi", gaussian_blur)
cv2.waitKey(0)
cv2.destroyAllWindows()

## median blur (it is good for salt and peeper noise )

median_blur = cv2.medianBlur(img, 5)
cv2.imshow("Messi", median_blur)
cv2.waitKey(0)
cv2.destroyAllWindows()

cv2.imshow("Messi", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
