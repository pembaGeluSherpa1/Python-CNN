```python
# 1, WAP to read an integer and check whether the number is even or odd using arithmetic operators and if-else.
num = int(input("Enter a number to check if it is even or odd: "))
if num%2==0:
    print(num,"is a even number")
else:
    print(num,"is a odd number")
```

```python
# 2, WAP to read two numbers and print the greater number using relational and logical operators.
num1, num2 = map(int, input("Enter 2 numbers to see the greater one:").split())
if num1>num2:
    print(num1,"is the greater number")
else:
    print(num2,"is a greater number")
```

```python
# 3, WAP to read a number and check whether it is positive, negative, or zero using if-elif-else.
num = int(input("Enter a number to check weather it is positive, negetive or zero:"))
if num == 0:
    print(num,"is zero")
elif num>0:
    print(num,"is positive")
else:
    print(num,"is negetive")
```

```python
# 4, WAP to read a number and print its multiplication table up to 10 using a for loop.
num = int(input("Enter a number to get its multiplication:"))
for i in range(1,11):
    print(f'{num} x {i} = {num*i}')
```
