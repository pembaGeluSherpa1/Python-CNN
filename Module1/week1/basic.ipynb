    - "numbers" → tuple

## for loop
- it repeatedly execute the given set of statement until the condition is meet
- entry control loop: while
- exit control loop: do while
- for loop 
- for each loop 

```python
# while loop

cnt = 0
while cnt<10:
    cnt+=1
    print(f"{cnt} Hello World!")
else:
    print("Java class")
```
