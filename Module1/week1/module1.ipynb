```python
age = int(input("Enter a age: "))
if age<18 and age>0:
    print(f'You are {age}, you are child')
elif age>18 and age<60:
    print(f'You are {age}, Adult')
elif age>60:
    print(f'You are {age}, Old')
else:
    print(f"Invalid age {age}")
```
