#### Bank Management System
- the simple bank management system is created using the oop(Object Oriented Programming)
- the banking operation we will perform are:-
  - Create account
  - Deposit cash
  - Withdraw cash
  - See account details

````python
class Customer():
    def __init__(self, name, age, address, contact):
        self.name = name
        self.age = age
        self.address = address
        self.contact = contact
    
    def show_detail(self):
        print("Name = " ,self.name)
        print("Age = ",self.age)
        print("Address = ",self.address)
        print("Contact = ",self.contact)
````

````python
class Bank(Customer):
    def __init__(self, name, age, address, contact):
        super().__init__(name, age, address, contact)
        self.balance = 0

    def deposit(self,amount):
        self.amount = amount
        self.balance += self.amount
        print("Your account balance is credited by RS:", self.amount)
        print("Your total balance is RS:",self.balance)
    
    def withdraw(self,amount):
        self.amount = amount
        if self.balance<self.amount:
            print("Insufficent balance")
            print("Your balance is RS:",self.balance)
        else:
            print("Cash Withdraw Successful....",self.amount)
            self.balance -= self.amount
            print("Your current balance is RS:",self.balance)
    
    def check_balance(self):
        print("Your current balance is RS:",self.balance)
````

````python
c1 = Bank("Pemba",21,"KTM",9812212212)
c1.show_detail()
````

````python
c1.deposit(10000)
````

````python
c1.withdraw(11000)
````

````python
c1.withdraw(9000)
````

````python
c1.check_balance()
````

#### Exception handling
``` Exception handling ```

- exception handling allows program to handle unexpected events
    - Missing input, missing field etc.
**To ensure system doesn't crash we use exception**
- Example:-
    - You enter a wrong password
    - User doesn't exist
    - Server/database down
- It avoid system crashes and display the message
- Syntax:
```
    try:
        ## code
    except exceptionName as e:
        ## display message 
```

````python
def demo_zero_division_error():
    try:
        result = 10/0 # divide by zero exception
    except ZeroDivisionError as z:
        print("Error encountered :",z)

def demo_value_error():
    try:
        number = int("not a number")
    except ValueError as e:
        print("Error encountered :",e)

def demo_index_error():
    arr = [1,2,3,4]
    try:
        element = arr[5]
    except IndexError as e:
        print("Error encountered :",e)

def demo_key_error():
    try:
        d = {"a":1,"b":2}
        value = d["c"]
    except KeyError as e:
        print("Error encountered :",e)

def demo_type_error():
    try:
        result = "2"+2
    except TypeError as e:
        print("Error encountered :",e)

if __name__ == "__main__": # idiom
    print("Divided by zero exception:")
    demo_zero_division_error()
    print("\nValue Error Exception :")
    demo_value_error()
    print("\nType Error Exception :")
    demo_type_error()
    print("\nKey Error Exception :")
    demo_key_error()
    print("\nIndex Error Exception :")
    demo_index_error()

````

#### file handling in python

- file handling is used to access data from the file, write into the file, etc

````python
try:
    ## writing to the file(overwrite the content)
    with open("demo.txt","w") as file:
        file.write("Hello, python file handling \n")
        file.write("Hello, this is a file handling operation \n")
        print("file written successfully!...")
except IOError as e:
    print("An IOError occurred :",e)

````

````python
try:
    ## reading from the file()
    with open("demo.txt","r") as file:
        content = file.read()
        print(content)
except IOError as e:
    print("An IOError occurred :",e)
````

````python
## append the file
try:
    ## apending the file()
    with open("demo.txt","a") as file:
        file.write("File being appended \n")
        print("file appended successfully!...")
except IOError as e:
    print("An IOError occurred :",e)
````

#### funtion recall
- create a class Shape
- Shape -> rectangle, square, circle,trangle
- function calculate area, perimeter

[Note: polymorphism(function overriding)]
[Abstraction ]

````python
class Shape():
    def calculate_area(self):
        pass
    
    def calculate_perimeter(self):
        pass

class Rectangle(Shape):
    def __init__(self, l,b):
        self.l = l
        self.b = b
    
    def calculate_perimeter(self):
        perimeter = 2*(self.l+self.b)
        print("perimeter of rectangle",perimeter)

    def calculate_area(self):
        area = self.l*self.b
        print("area of rectangle", area)

class Square(Shape):
    def __init__(self, l):
        self.l = l
    
    def calculate_perimeter(self):
        perimeter = 4*self.l
        print("perimeter of rectangle",perimeter)

    def calculate_area(self):
        area = self.l*self.l
        print("area of rectangle", area)

class Circle(Shape):
    def __init__(self, r):
        self.r = r
    
    def calculate_perimeter(self):
        perimeter = 2*22/7*self.r
        print("perimeter of rectangle",perimeter)

    def calculate_area(self):
        area = 22/7*self.r*self.r
        print("area of rectangle", area)

class Trangle(Shape):
    def __init__(self, a,b,c):
        self.a = a
        self.b = b
        self.c = c
        self.h = 0
    
    def calculate_perimeter(self):
        perimeter = self.a+self.b+self.c
        print("perimeter of rectangle",perimeter)

    def calculate_area(self,h):
        self.h = h
        area = 1/2*self.b*self.h
        print("area of rectangle", area)
````

````python
s1 = Rectangle(2,3)
s1.calculate_area()
s1.calculate_perimeter()
````

````python
s2 = Trangle(2,3,4)
s2.calculate_area(5)
s2.calculate_perimeter()
````
