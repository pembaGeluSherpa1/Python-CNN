### statics using numpy

````python
arra = np.array([10,20,30,40,50,60])
print("mean = ",np.mean(arra))
print("median = ",np.median(arra))
print("sum = ",np.sum(arra))
print("standard deviation = ",np.std(arra))
print("min = ",np.min(arra))
print("max = ",np.max(arra))
print("variance = ",np.var(arra))
print("square root = ",np.sqrt(np.max(arra)))
print("percentile 25 = ",np.percentile(arra,25))
print("percentile 75 = ",np.percentile(arra,75))
print("quartile = ",np.percentile(arra,[25,50,75]))
````

### matrix computation

````python
A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])

print(f"Sum of matrix a and b \n {A+B}")
print("---------------------------------------------")

print(f"Sub of matrix a and b \n {A-B}")
print("---------------------------------------------")

print(f"multiplication of matrix a and b \n {A*B}")
print("---------------------------------------------")

print("dot product :\n ",np.dot(A,B))
print("---------------------------------------------")

print(f"Transpose :\n{A.T}")
print("---------------------------------------------")

print(f"inverse :\n{np.linalg.inv(A)}")
print("----------------------------------------------")

print(f"Determinate :\n{np.linalg.det(A)}")
````

### solving the linear equation 

``` 
2x + 3y = 8
5x + 7y = 19
``` 

````python
cofficient = np.array([[2,3],[5,7]])
constants = np.array([8,19])
inverse_cofficient = np.linalg.inv(cofficient)
print(np.dot(inverse_cofficient,constants))
````

````python
solution = np.linalg.solve(cofficient,constants)
print(solution)
````
